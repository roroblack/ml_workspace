```python
text = """Beneath it were the words: "Stay Hungry. Stay Foolish."
It was their farewell message as they signed off.
And I have always wished that for myself.
And now, as you graduate to begin anew, I wish that for you."""
```

## 정제 (Cleaning)

특수문자 제거

```python
import string

print(string.punctuation)
```

```python
import re
# 파이썬이 제공하는 정규표현식 제공 표준 라이브러리

cleaned_text = re.sub('[^\w\.\s]', '', text)
print (cleaned_text)
```

```python
cleaned_text = re.sub('\n', '', cleaned_text)
print(cleaned_text)
```

## 토큰화 (Tokenizatioin)

```python
import nltk
nltk.download('punkt')

nltk.download('punkt_tab')
```

```python
sent_tokens = nltk.sent_tokenize(cleaned_text)
print(sent_tokens)
```

```python
tokens = nltk.word_tokenize(cleaned_text)

tokens
```

## 불용어 제거 (Stopword Elimination)

```python
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
stop = stopwords.words('english')
print(stop)
```

```python
tokens = [token for token in tokens if token not in stop]
print(tokens)
```

## 소문자화(lower capitalization)

```python
print('소문자화 수행전 : ', tokens)
tokens_lower = [token.lower() for token in tokens]
print('소문자화 수행후 : ', tokens_lower)
```
