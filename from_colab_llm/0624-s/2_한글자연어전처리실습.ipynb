# 한글 자연어 전처리

```python
text = '''그 밑에는 "계속 배고픔을 느끼세요. 계속 바보로 남으세요" 라는 문구가 새겨져 있었습니다.
그들이 전한 마지막 인사말이었습니다. 계속 배고픔을 느끼세요. 계속 바보로 남으세요.
그리고 저는 항상 제 자신이 그렇길 바랬습니다. 이제는 졸업을 하고 새로운 출발을 하는
여러분에게 바라는 바 입니다.'''
```

## 정재 (Cleaning)

영어와 같음. 특수문자 제거

```python
import string
import re

cleaned_text = re.sub('[^\w\s]', '', text) # 단어와 공백 빼고 다 제거
print(cleaned_text)

```

```python
cleaned_text = re.sub('\n', '', text)
print(cleaned_text)
```

```python
cleaned_text = re.sub('\s', '', text)
print(cleaned_text)
```

```python
non_space = re.sub('\s', '', text)
print(non_space)
```

```python
from pykospacing import Spacing
spacing = Spacing()
new_sent = Spacing(non_space)
print(new_sent)
```

## 토큰화 (Tokenization)


```python
!pip install kss
```

```python

import kss
sent_tokens = kss.split_sentences(cleaned_text)
print(sent_tokens)
```

```python
def tokenizer(words):
    tokens = words.split()
    return tokens

tokens = tokenizer(cleaned_text)
print(tokens)
```

## 불용어 제거 (Stopword Elimination)

```python
from google.colab import files
import os

data_dir = 'data'
if not os.path.exists(data_dir):
    os.mkdir(data_dir)

os.chdir(data_dir)
files.upload()
os.chdir('..')
```

```python
import pandas as pd
stop = pd.read_csv('./data/stopword_dict.csv')
print(stop[:10])
```

```python
print(list(stop.stopword))
```

```python
# 불용어 제거
tokens = [
    token for token in tokens
    if token not in list(stop.stopword)
]

print(tokens)
```
