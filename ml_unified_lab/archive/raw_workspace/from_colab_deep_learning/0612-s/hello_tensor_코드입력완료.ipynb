## 1. TensorFlow와 Keras 개념

TensorFlow는 딥러닝 모델을 만들고 학습시키기 위한 대표적인 라이브러리입니다.  
Keras는 TensorFlow 안에서 신경망 모델을 더 쉽게 만들 수 있도록 도와주는 고수준 API입니다.

이 예제에서는 `tf.keras`를 사용하여 MNIST 손글씨 숫자 분류 모델을 만듭니다.  
MNIST 데이터셋은 0부터 9까지의 손글씨 숫자 이미지로 구성되어 있으며, 딥러닝 입문에서 가장 많이 사용하는 기본 데이터셋입니다.

```python
# TensorFlow 라이브러리를 tf라는 짧은 이름으로 불러옵니다.
# TensorFlow는 텐서 연산, 신경망 구성, 모델 학습 등을 지원하는 딥러닝 라이브러리입니다.
import tensorflow as tf

# NumPy는 배열 계산을 쉽게 처리하기 위한 파이썬 수치 계산 라이브러리입니다.
# 여기서는 예측 결과를 보기 좋게 처리할 때 사용합니다.
import numpy as np

# Matplotlib은 그래프와 이미지를 화면에 출력하기 위한 시각화 라이브러리입니다.
# 여기서는 MNIST 손글씨 이미지를 확인하고 학습 결과 그래프를 그릴 때 사용합니다.
import matplotlib.pyplot as plt

# os는 운영체제 관련 설정을 다루는 표준 라이브러리입니다.
# TensorFlow 로그 출력 수준을 조정할 때 사용합니다.
import os

# TensorFlow의 불필요한 안내 로그를 줄이기 위한 설정입니다.
# Colab에서는 필수는 아니지만 출력 화면을 깔끔하게 유지하는 데 도움이 됩니다.
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"

# 실험 결과를 가능한 한 재현 가능하게 만들기 위해 난수 시드를 고정합니다.
# 같은 코드를 다시 실행했을 때 비슷한 학습 결과가 나오도록 돕습니다.
tf.keras.utils.set_random_seed(42)

# 현재 Colab 또는 실행 환경에 설치된 TensorFlow 버전을 출력합니다.
# 버전을 확인하면 코드가 어떤 TensorFlow 환경에서 실행되는지 알 수 있습니다.
print("TensorFlow 버전:", tf.__version__)
```

## 2. TensorFlow 텐서 생성과 출력

`tf.constant()`는 값이 고정된 텐서를 만드는 함수입니다.  
텐서는 딥러닝에서 사용하는 기본 데이터 단위입니다. 숫자 하나, 벡터, 행렬, 이미지 데이터 모두 텐서로 표현할 수 있습니다.

`tf.print()`는 TensorFlow 텐서를 출력할 때 사용하는 함수입니다.  
일반적인 `print()`도 사용할 수 있지만, TensorFlow 연산 결과를 안정적으로 출력할 때는 `tf.print()`가 자주 사용됩니다.

```python
# 문자열 데이터를 TensorFlow 텐서로 생성합니다.
# tf.constant()는 변경되지 않는 고정값 텐서를 만들 때 사용합니다.
msg = tf.constant("hello tensor!")

# TensorFlow 방식으로 텐서 값을 출력합니다.
# 출력 결과는 b'hello world!' 또는 hello world! 형태로 표시될 수 있습니다.
tf.print(msg)
```

## 3. MNIST 데이터셋 불러오기

`tf.keras.datasets.mnist.load_data()`는 TensorFlow가 제공하는 MNIST 데이터를 불러오는 함수입니다.

불러온 데이터는 다음과 같이 네 부분으로 나누어집니다.

- `X_train`: 학습용 이미지 데이터
- `Y_train`: 학습용 정답 라벨
- `X_test`: 평가용 이미지 데이터
- `Y_test`: 평가용 정답 라벨

학습용 데이터는 모델이 공부할 때 사용하고, 평가용 데이터는 공부가 끝난 모델이 새로운 문제를 얼마나 잘 맞히는지 확인할 때 사용합니다.

```python
# TensorFlow Keras에 포함된 MNIST 데이터셋 모듈을 mnist라는 변수에 저장합니다.
# MNIST는 0~9 숫자 손글씨 이미지 데이터셋입니다.
mnist = tf.keras.datasets.mnist

# load_data() 함수는 MNIST 데이터를 자동으로 다운로드하고 메모리에 불러옵니다.
# 반환값은 ((학습 이미지, 학습 정답), (평가 이미지, 평가 정답)) 구조입니다.
(X_train, Y_train), (X_test, Y_test) = mnist.load_data()

# 학습용 입력 데이터의 모양을 출력합니다.
# 일반적으로 (60000, 28, 28) 형태이며, 28x28 이미지가 60000장 있다는 뜻입니다.
print("학습용 입력 데이터 모양:", X_train.shape)

# 학습용 출력 데이터의 모양을 출력합니다.
# 일반적으로 (60000,) 형태이며, 각 이미지의 정답 숫자가 60000개 있다는 뜻입니다.
print("학습용 출력 데이터 모양:", Y_train.shape)

# 평가용 입력 데이터의 모양을 출력합니다.
# 일반적으로 (10000, 28, 28) 형태이며, 평가용 이미지가 10000장 있다는 뜻입니다.
print("평가용 입력 데이터 모양:", X_test.shape)

# 평가용 출력 데이터의 모양을 출력합니다.
# 일반적으로 (10000,) 형태이며, 평가용 정답 숫자가 10000개 있다는 뜻입니다.
print("평가용 출력 데이터 모양:", Y_test.shape)
```

MNIST 이미지는 28×28 크기의 흑백 이미지입니다.  
각 픽셀은 0부터 255 사이의 값을 가집니다.

- 0에 가까울수록 검은색
- 255에 가까울수록 흰색

`plt.imshow()`는 배열 형태의 이미지 데이터를 실제 이미지처럼 보여주는 함수입니다.  
`cmap='gray'`는 흑백 이미지로 출력하겠다는 의미입니다.

## 5. 이미지 데이터 정규화

딥러닝 모델은 입력값의 범위가 너무 크면 학습이 불안정해질 수 있습니다.  
MNIST 원본 픽셀값은 0부터 255 사이이므로, 이를 255로 나누어 0부터 1 사이의 값으로 변환합니다.

이 과정을 정규화 또는 스케일링이라고 합니다.

정규화 전:

- 픽셀값 범위: 0 ~ 255

정규화 후:

- 픽셀값 범위: 0.0 ~ 1.0

```python
# 학습 이미지 데이터를 255.0으로 나누어 0~1 범위의 실수 데이터로 변환합니다.
# 신경망은 일반적으로 작은 범위의 숫자를 입력받을 때 더 안정적으로 학습됩니다.
X_train = X_train.astype("float32") / 255.0

# 평가 이미지 데이터도 학습 데이터와 같은 방식으로 정규화합니다.
# 학습 데이터와 평가 데이터는 반드시 같은 전처리 규칙을 적용해야 합니다.
X_test = X_test.astype("float32") / 255.0

# 정규화 후 첫 번째 학습 이미지를 다시 출력합니다.
# 화면상 이미지는 비슷해 보이지만 내부 픽셀값은 0~1 범위로 바뀌었습니다.
plt.imshow(X_train[0], cmap="gray")

# 정규화 후 이미지 제목을 표시합니다.
plt.title(f"정규화 후 첫 번째 학습 이미지 / 정답: {Y_train[0]}")

# 축 표시를 제거합니다.
plt.axis("off")

# 이미지를 화면에 출력합니다.
plt.show()

# 정규화 후 첫 번째 이미지의 픽셀값 배열을 출력합니다.
# 값들이 0~1 사이의 실수로 바뀐 것을 확인할 수 있습니다.
print("정규화 후 첫 번째 학습용 데이터 입력값:")
print(X_train[0])

# 정규화 후 최소값과 최대값을 출력합니다.
# 정상적으로 정규화되었다면 최소값은 0.0, 최대값은 1.0에 가깝게 나옵니다.
print("정규화 후 학습 데이터 최소값:", X_train.min())
print("정규화 후 학습 데이터 최대값:", X_train.max())
```

## 6. DNN 모델 설계

이 예제에서는 `tf.keras.Sequential`을 사용합니다.  
`Sequential`은 층을 순서대로 쌓아 만드는 가장 기본적인 신경망 모델 클래스입니다.

모델 구조는 다음과 같습니다.

1. `Input`: 입력 이미지의 모양 지정
2. `Flatten`: 28×28 이미지를 784개의 숫자로 펼침
3. `Dense`: 완전연결층으로 특징 학습
4. `Dropout`: 과적합 방지
5. `Dense`: 10개 숫자 클래스에 대한 확률 출력

마지막 출력층은 숫자 0부터 9까지 총 10개 클래스를 예측해야 하므로 뉴런 수가 10개입니다.  
활성화 함수는 `softmax`를 사용하여 각 숫자일 확률을 출력합니다.

```python
# from tensorflow.keras import Sequential
# from tensorflow.keras.layers import Input, Flatten, Dense

# Sequential 안에 리스트로 구성
model = tf.keras.Sequential([

    # Input은 모델이 받을 데이터의 형태를 명확하게 지정하는 입력층입니다.
    # MNIST 이미지는 28행 28열의 2차원 배열이므로 input_shape를 (28, 28)로 지정합니다.
    tf.keras.layers.Input(shape=(28, 28)),

    # Flatten은 2차원 이미지 데이터를 1차원 벡터로 펼치는 층입니다.
    # 28 x 28 이미지는 784개의 숫자로 변환됩니다.
    tf.keras.layers.Flatten(),

    # Dense는 모든 입력 노드와 모든 출력 노드가 연결되는 완전연결층입니다.
    # 128은 뉴런 개수이며, relu는 음수는 0으로 만들고 양수는 그대로 통과시키는 활성화 함수입니다.
    tf.keras.layers.Dense(128, activation='relu'),

    # Dropout은 학습 중 일부 뉴런을 무작위로 꺼서 과적합을 줄이는 층입니다.
    # 0.2는 학습할 때 전체 뉴런 중 20%를 임시로 사용하지 않겠다는 의미입니다.
    tf.keras.layers.Dropout(0.2),

    # 마지막 Dense 층은 출력층입니다.
    # MNIST는 0~9까지 10개 숫자 중 하나를 분류하므로 뉴런 수를 10개로 지정합니다.
    # softmax는 10개 클래스 각각에 대한 확률값을 출력합니다.
    tf.keras.layers.Dense(10, activation='softmax')
])

# model.summary()는 모델의 층 구조, 출력 모양, 학습 가능한 파라미터 수를 요약해서 보여줍니다.
# 모델이 의도한 대로 구성되었는지 확인할 때 매우 중요합니다.
model.summary()
```

## 7. 모델 컴파일

모델을 학습하기 전에는 `compile()`로 학습 환경을 설정해야 합니다.

`compile()`에서 주로 지정하는 항목은 다음과 같습니다.

- `optimizer`: 가중치를 어떻게 수정할지 결정하는 최적화 알고리즘
- `loss`: 예측값과 정답의 차이를 계산하는 손실 함수
- `metrics`: 학습 중 확인할 평가 지표

이 예제에서는 다음 설정을 사용합니다.

- Optimizer: `adam`
- Loss: `sparse_categorical_crossentropy`
- Metric: `accuracy`

MNIST 정답 라벨은 0, 1, 2처럼 정수 형태이므로 `sparse_categorical_crossentropy`를 사용합니다.

```python
# compile()은 모델 학습에 필요한 환경을 설정하는 함수입니다.
model.compile(

    # optimizer는 손실을 줄이기 위해 가중치를 업데이트하는 알고리즘입니다.
    # adam은 실무와 입문 예제에서 널리 사용되는 안정적인 최적화 알고리즘입니다.
    optimizer = 'adam',

    # loss는 모델의 예측값과 실제 정답 사이의 차이를 계산하는 함수입니다.
    # sparse_categorical_crossentropy는 정답이 원-핫 인코딩이 아닌 정수 라벨일 때 사용하는 다중분류 손실함수입니다.
    loss = 'sparse_categorical_crossentropy',

    # metrics는 학습 과정에서 추가로 확인할 평가 지표입니다.
    # accuracy는 전체 데이터 중 모델이 정답을 맞힌 비율입니다.
    metrics = ['accuracy']
)
```

## 8. 모델 훈련

`fit()`은 모델을 실제로 학습시키는 함수입니다.  
모델은 학습 데이터를 여러 번 반복해서 보면서 가중치를 조정합니다.

주요 파라미터는 다음과 같습니다.

- `X_train`: 학습 입력 이미지
- `Y_train`: 학습 정답 라벨
- `epochs`: 전체 학습 데이터를 몇 번 반복해서 학습할지 지정
- `batch_size`: 한 번에 몇 개의 데이터를 묶어서 학습할지 지정
- `validation_split`: 학습 데이터 일부를 검증용으로 분리

검증 데이터는 모델이 학습 중 과적합되고 있는지 확인하는 데 사용됩니다.

```python
# fit() 함수는 모델을 학습시키는 함수입니다.
# 학습 과정에서 손실값과 정확도가 출력됩니다.
# 학습 결과(로그)를 시각화하기 위해 history 변수에 할당합니다.
history = model.fit(

    # X_train은 학습에 사용할 입력 이미지 데이터입니다.
    X_train,

    # Y_train은 각 이미지에 해당하는 정답 숫자 라벨입니다.
    Y_train,

    # epochs는 전체 학습 데이터를 몇 번 반복해서 학습할지 지정합니다.
    # 5는 전체 학습 데이터를 5번 반복해서 모델에 보여준다는 뜻입니다.
    epochs = 5,

    # batch_size는 한 번의 가중치 업데이트에 사용할 데이터 개수입니다.
    # 128개씩 묶어서 학습하면 속도와 안정성의 균형을 잡을 수 있습니다.
    batch_size = 128,

    # validation_split은 학습 데이터 중 일부를 검증용으로 자동 분리하는 옵션입니다.
    # 0.1은 학습 데이터의 10%를 검증 데이터로 사용한다는 의미입니다.
    validation_split = 0.1,

    # verbose=1은 학습 진행 상황을 자세히 출력하겠다는 의미입니다.
    verbose=1
)
```

## 9. 학습 과정 시각화

`history` 객체에는 epoch별 손실값과 정확도 변화가 저장됩니다.  
이 값을 그래프로 확인하면 모델이 학습되면서 성능이 좋아지는지, 과적합이 발생하는지 확인할 수 있습니다.

일반적으로 다음과 같이 해석합니다.

- 학습 손실과 검증 손실이 함께 감소하면 정상적으로 학습 중입니다.
- 학습 정확도는 오르는데 검증 정확도가 떨어지면 과적합 가능성이 있습니다.

```python
# history.history에는 학습 과정에서 기록된 손실값과 정확도가 딕셔너리 형태로 저장됩니다.
# 예: history.history["loss"], history.history["accuracy"] 등
print(history.history.keys())

# 학습 손실값을 선 그래프로 출력합니다.
plt.plot(history.history["loss"], label="train_loss")

# 검증 손실값을 선 그래프로 출력합니다.
plt.plot(history.history["val_loss"], label="val_loss")

# 그래프 제목을 지정합니다.
plt.title("Epoch별 손실값 변화")

# x축 이름을 지정합니다.
plt.xlabel("Epoch")

# y축 이름을 지정합니다.
plt.ylabel("Loss")

# 각 선이 무엇을 의미하는지 범례를 표시합니다.
plt.legend()

# 그래프를 화면에 출력합니다.
plt.show()

# 학습 정확도를 선 그래프로 출력합니다.
plt.plot(history.history["accuracy"], label="train_accuracy")

# 검증 정확도를 선 그래프로 출력합니다.
plt.plot(history.history["val_accuracy"], label="val_accuracy")

# 그래프 제목을 지정합니다.
plt.title("Epoch별 정확도 변화")

# x축 이름을 지정합니다.
plt.xlabel("Epoch")

# y축 이름을 지정합니다.
plt.ylabel("Accuracy")

# 범례를 표시합니다.
plt.legend()

# 그래프를 화면에 출력합니다.
plt.show()
```

## 10. 모델 평가

`evaluate()`는 학습이 끝난 모델을 평가용 데이터로 검증하는 함수입니다.  
평가용 데이터는 모델 훈련에 직접 사용되지 않았기 때문에, 모델이 새로운 데이터에 대해 얼마나 잘 예측하는지 확인할 수 있습니다.

평가 결과로는 손실값과 정확도가 출력됩니다.

```python
# evaluate()는 테스트 데이터를 사용하여 모델 성능을 평가하는 함수입니다.
# test_loss는 테스트 데이터에서 계산된 손실값입니다.
# test_accuracy는 테스트 데이터에서 계산된 정확도입니다.
test_loss, test_accuracy = model.evaluate(

    X_test,
    Y_test,

    # verbose=1은 평가 진행 상황을 출력하겠다는 의미입니다.
    verbose=1
)

# 테스트 손실값을 출력합니다.
print("테스트 손실값:", test_loss)

# 테스트 정확도를 출력합니다.
print("테스트 정확도:", test_accuracy)
```

## 11. 모델 예측

`predict()`는 학습된 모델을 사용하여 새로운 입력 데이터의 정답을 예측하는 함수입니다.  
여기서는 테스트 데이터 중 첫 번째 이미지를 선택하여 모델이 어떤 숫자로 예측하는지 확인합니다.

모델의 출력은 10개 숫자에 대한 확률값입니다.  
가장 큰 확률을 가진 인덱스가 모델이 예측한 숫자가 됩니다.

```python
# 테스트 데이터 중 첫 번째 이미지를 선택합니다.
# X_test[0]의 모양은 (28, 28)입니다.
sample_image = X_test[0]

# 모델은 여러 개의 데이터를 한 번에 입력받는 구조를 기대합니다.
# 따라서 reshape(1, 28, 28)을 사용하여 데이터 개수 차원을 추가합니다.
# 1은 이미지 1장을 의미합니다.
pick = sample_image.reshape(1, 28, 28)

# predict()는 입력 이미지에 대한 모델의 예측 확률을 반환합니다.
# 결과는 10개 숫자 클래스에 대한 확률 배열입니다.
pred = model.predict(pick)

# np.argmax()는 배열에서 가장 큰 값의 인덱스를 반환합니다.
# softmax 출력에서 가장 큰 확률의 인덱스가 모델이 예측한 숫자입니다.
answer = np.argmax(pred, axis=1)

# 예측에 사용한 이미지를 화면에 출력합니다.
plt.imshow(sample_image, cmap="gray")

# 제목에 모델 예측값과 실제 정답을 함께 표시합니다.
plt.title(f"예측값: {answer[0]} / 실제 정답: {Y_test[0]}")

# 축 표시를 제거합니다.
plt.axis("off")

# 이미지를 화면에 출력합니다.
plt.show()

# 모델이 출력한 원본 확률값 배열을 출력합니다.
print("인공신경망 추측 결과 원본 확률값:")
print(pred)

# 모델이 최종적으로 선택한 숫자를 출력합니다.
print("인공신경망 추측 결과 해석:", answer[0])

# 실제 정답 라벨을 출력합니다.
print("정답:", Y_test[0])
```

## 12. 여러 개의 테스트 이미지 예측 결과 확인

한 장의 이미지만 확인하면 모델의 전체적인 예측 성능을 직관적으로 이해하기 어렵습니다.  
따라서 여러 개의 테스트 이미지를 한 번에 예측하고, 예측값과 실제 정답을 함께 확인합니다.

이 과정은 모델이 어떤 숫자를 잘 맞히고 어떤 숫자를 헷갈리는지 확인하는 데 도움이 됩니다.

```python
# 테스트 데이터에서 앞쪽 10개의 이미지를 선택합니다.
# selected_images의 모양은 (10, 28, 28)입니다.
selected_images = X_test[:10]

# 선택한 10개 이미지에 대해 모델 예측을 수행합니다.
# predict() 결과는 각 이미지마다 10개 클래스 확률을 반환합니다.
pred_probs = model.predict(selected_images)

# 각 이미지별로 가장 확률이 높은 숫자 클래스를 선택합니다.
pred_labels = np.argmax(pred_probs, axis=1)

# 10개의 이미지를 한 줄에 출력하기 위해 그래프 크기를 설정합니다.
plt.figure(figsize=(15, 3))

# 10개의 이미지를 반복하면서 하나씩 출력합니다.
for i in range(10):

    # 1행 10열 구조의 그래프 중 i+1번째 위치를 선택합니다.
    plt.subplot(1, 10, i + 1)

    # i번째 테스트 이미지를 흑백으로 출력합니다.
    plt.imshow(selected_images[i], cmap="gray")

    # 제목에 예측값과 실제 정답을 표시합니다.
    plt.title(f"P:{pred_labels[i]}\nT:{Y_test[i]}")

    # 축 표시를 제거합니다.
    plt.axis("off")

# 여러 이미지가 겹치지 않도록 배치를 자동 조정합니다.
plt.tight_layout()

# 그래프를 화면에 출력합니다.
plt.show()
```

## 13. 주요 클래스와 함수 사용 설명

### `tf.constant()`
고정된 값을 가진 TensorFlow 텐서를 만듭니다.  
문자열, 숫자, 배열 등을 텐서로 변환할 수 있습니다.

### `tf.keras.datasets.mnist.load_data()`
MNIST 손글씨 숫자 데이터셋을 불러옵니다.  
학습 데이터와 평가 데이터를 자동으로 나누어 제공합니다.

### `tf.keras.Sequential`
층을 순서대로 쌓아 신경망 모델을 만드는 클래스입니다.  
입문용 모델 구성에서 가장 많이 사용됩니다.

### `tf.keras.layers.Input`
모델이 입력받을 데이터의 모양을 지정합니다.  
최신 Keras에서는 `Flatten(input_shape=...)`보다 `Input(shape=...)`을 먼저 두는 방식이 더 명확합니다.

### `tf.keras.layers.Flatten`
2차원 또는 3차원 데이터를 1차원 벡터로 펼칩니다.  
이미지를 Dense 층에 넣기 전에 자주 사용합니다.

### `tf.keras.layers.Dense`
완전연결층입니다.  
이전 층의 모든 값이 현재 층의 모든 뉴런과 연결됩니다.

### `tf.keras.layers.Dropout`
학습 중 일부 뉴런을 무작위로 꺼서 과적합을 줄입니다.  
모델이 특정 뉴런에만 의존하지 않도록 돕습니다.

### `model.compile()`
모델 학습 방식을 설정합니다.  
최적화 알고리즘, 손실 함수, 평가 지표를 지정합니다.

### `model.fit()`
모델을 실제로 학습시킵니다.  
입력 데이터와 정답 데이터를 이용하여 가중치를 반복적으로 수정합니다.

### `model.evaluate()`
학습이 끝난 모델의 성능을 테스트 데이터로 평가합니다.  
손실값과 정확도를 확인할 수 있습니다.

### `model.predict()`
새로운 입력 데이터에 대한 예측값을 계산합니다.  
분류 문제에서는 클래스별 확률을 출력합니다.
