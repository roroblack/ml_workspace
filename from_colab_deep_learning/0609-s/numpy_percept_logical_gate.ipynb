```python
# ============================================================
# 1. AND 게이트 구현
# ============================================================

import numpy as np

def AND(x1, x2):
    # 입력값 x1, x2를 NumPy 배열로 생성한다.
    x = np.array([x1, x2])

    # AND 게이트를 만족하도록 설정한 가중치이다.
    # 두 입력이 모두 1일 때만 임계값을 넘도록 한다.
    w = np.array([0.5, 0.5])

    # 편향값이다.
    # 입력값의 가중합이 너무 쉽게 1이 되지 않도록 기준점을 낮춘다.
    b = -0.7

    # 입력값과 가중치를 곱한 뒤 모두 더하고 편향을 더한다.
    tmp = np.sum(x * w) + b

    # 계산 결과가 0 이하이면 출력 0
    if tmp <= 0:
        return 0
    # 계산 결과가 0보다 크면 출력 1
    else:
        return 1


print(AND(0, 0))
print(AND(1, 0))
print(AND(0, 1))
print(AND(1, 1))
```

```python
# ============================================================
# 2. NAND 게이트 구현
# ============================================================

import numpy as np

def NAND(x1, x2):
    # 입력값을 NumPy 배열로 생성한다.
    x = np.array([x1, x2])

    # NAND는 AND의 반대 동작을 하므로 음수 가중치를 사용한다.
    w = np.array([-0.2, -0.2])

    # 편향값이다.
    # 두 입력이 모두 1일 때만 결과가 0 이하가 되도록 조정한다.
    b = 0.3

    # 입력값과 가중치의 곱을 모두 더한 뒤 편향을 더한다.
    tmp = np.sum(x * w) + b

    # 계산 결과가 0 이하이면 출력 0
    if tmp <= 0:
        return 0
    # 계산 결과가 0보다 크면 출력 1
    else:
        return 1


print(NAND(0, 0))
print(NAND(1, 0))
print(NAND(0, 1))
print(NAND(1, 1))
```

```python
# ============================================================
# 3. OR 게이트 구현
# ============================================================

import numpy as np

def OR(x1, x2):
    # 입력값 x1, x2를 NumPy 배열로 생성한다.
    x = np.array([x1, x2])

    # OR 게이트를 만족하도록 설정한 가중치이다.
    # 하나의 입력만 1이어도 임계값을 넘도록 한다.
    w = np.array([0.7, 0.7])

    # 편향값이다.
    # 두 입력이 모두 0일 때는 출력이 0이 되도록 한다.
    b = -0.4

    # 입력값과 가중치를 곱한 뒤 합산하고 편향을 더한다.
    tmp = np.sum(x * w) + b

    # 계산 결과가 0 이하이면 출력 0
    if tmp <= 0:
        return 0
    # 계산 결과가 0보다 크면 출력 1
    else:
        return 1


print(OR(0, 0))
print(OR(1, 0))
print(OR(0, 1))
print(OR(1, 1))
```

```python
# ============================================================
# 4. 학습이 아닌 무작위 탐색으로 w, b 찾기
# ============================================================

import numpy as np

def check(x1, x2, y):
    # 올바른 w1, w2, b를 찾을 때까지 반복한다.
    while True:
        # 정답 개수를 저장하는 변수이다.
        cnt = 0

        # 평균 0, 표준편차 1인 정규분포에서 w1을 무작위 생성한다.
        w1 = np.random.normal()

        # 평균 0, 표준편차 1인 정규분포에서 w2를 무작위 생성한다.
        w2 = np.random.normal()

        # 평균 0, 표준편차 1인 정규분포에서 편향 b를 무작위 생성한다.
        b = np.random.normal()

        # 4개의 입력 조합을 하나씩 검사한다.
        for i in range(4):
            # 현재 입력값 x1[i], x2[i]를 NumPy 배열로 생성한다.
            x = np.array([x1[i], x2[i]])

            # 현재 무작위로 생성된 가중치 w1, w2를 배열로 생성한다.
            w = np.array([w1, w2])

            # 퍼셉트론 계산식: 입력 * 가중치 + 편향
            tmp = np.sum(x * w) + b

            # 계산 결과가 0 이하이면 출력 0
            if tmp <= 0:
                result = 0
            # 계산 결과가 0보다 크면 출력 1
            else:
                result = 1

            # 예측 결과가 실제 정답 y[i]와 같으면 정답 개수를 1 증가시킨다.
            if y[i] == result:
                cnt += 1

        # 4개의 입력 조합을 모두 맞히면 현재 w1, w2, b를 출력하고 반복을 종료한다.
        if cnt == 4:
            print("w1 :", w1)
            print("w2 :", w2)
            print("b  :", b)
            break
```

```python
# ============================================================
# 5. AND, NAND, OR용 입력 데이터와 정답 설정
# ============================================================

def AND_random_search():
    # AND 게이트 입력 데이터이다.
    x1 = [0, 1, 0, 1]
    x2 = [0, 0, 1, 1]

    # AND 게이트 정답이다.
    # 두 입력이 모두 1일 때만 1이다.
    y = [0, 0, 0, 1]

    # check 함수로 AND를 만족하는 w1, w2, b를 찾는다.
    check(x1, x2, y)


def NAND_random_search():
    # NAND 게이트 입력 데이터이다.
    x1 = [0, 1, 0, 1]
    x2 = [0, 0, 1, 1]

    # NAND 게이트 정답이다.
    # 두 입력이 모두 1일 때만 0이다.
    y = [1, 1, 1, 0]

    # check 함수로 NAND를 만족하는 w1, w2, b를 찾는다.
    check(x1, x2, y)


def OR_random_search():
    # OR 게이트 입력 데이터이다.
    x1 = [0, 1, 0, 1]
    x2 = [0, 0, 1, 1]

    # OR 게이트 정답이다.
    # 하나 이상의 입력이 1이면 1이다.
    y = [0, 1, 1, 1]

    # check 함수로 OR를 만족하는 w1, w2, b를 찾는다.
    check(x1, x2, y)
```

```python
# ============================================================
# 6. 무작위 탐색 실행
# ============================================================

print("AND")
AND_random_search()

print()

print("NAND")
NAND_random_search()

print()

print("OR")
OR_random_search()
```
